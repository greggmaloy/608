```python
import pandas as pd
import numpy as np
import folium
import requests
from folium.features import GeoJsonPopup

# ------------------------------------------------
# 1. Define which columns are energy sources
# ------------------------------------------------
# Map friendly labels -> column names in merged_df
energy_cols = {
    "Coal": "coal_production_2023",
    "Natural Gas": "ng_production_2023",
    "Petroleum": "petroleum_production_2023",      # change if your column name differs
    "Nuclear": "nuclear_production_2023",
    "Renewable": "renewable_production_2023"
}

# (Optional) make sure these columns are numeric
for col in ["total_production_2023", "total_consumption_2023", "total_ratio"] + list(energy_cols.values()):
    if col in merged_df.columns:
        merged_df[col] = pd.to_numeric(merged_df[col], errors="coerce")

# ------------------------------------------------
# 2. Build an HTML popup field for each state
# ------------------------------------------------
def make_popup_html(row):
    # header + totals
    parts = [
        f"<b>{row.name}</b>",
        f"Total Production: {row['total_production_2023']:,.2f} Btu",
        f"Total Consumption: {row['total_consumption_2023']:,.2f} Btu",
        f"Dependency Ratio: {row['total_ratio']*100:.1f}%"
    ]

    # energy mix
    parts.append("Energy Mix (%):")
    mix_lines = []
    total_prod = row["total_production_2023"]

    if pd.notna(total_prod) and total_prod > 0:
        for label, col in energy_cols.items():
            if col in row.index and pd.notna(row[col]):
                share = row[col] / total_prod * 100
                mix_lines.append(f"- {label}: {share:.1f}%")

    if mix_lines:
        parts.extend(mix_lines)

    # join with <br> for HTML line breaks
    return "<br>".join(parts)

# Use state code as index
state_data = merged_df.set_index("State").copy()
state_data["popup_html"] = state_data.apply(make_popup_html, axis=1)

# ------------------------------------------------
# 3. Load GeoJSON and attach popup_html + ratio
# ------------------------------------------------
geo_url = "https://raw.githubusercontent.com/python-visualization/folium/master/examples/data/us-states.json"
us_states = requests.get(geo_url).json()

for feature in us_states["features"]:
    st = feature["id"]
    if st in state_data.index:
        row = state_data.loc[st]
        feature["properties"].update({
            "popup_html": row["popup_html"],
            "total_ratio_capped": row["total_ratio_capped"]
        })

# ------------------------------------------------
# 4. Make the choropleth + popup
# ------------------------------------------------
m = folium.Map(location=[37.8, -96], zoom_start=4, tiles="cartodbpositron")

folium.Choropleth(
    geo_data=us_states,
    data=merged_df,
    columns=["State", "total_ratio_capped"],
    key_on="feature.id",
    fill_color="YlOrRd",
    fill_opacity=0.8,
    line_opacity=0.4,
    nan_fill_color="lightgray",
    legend_name="Total Energy Dependency Ratio (Consumption ÷ Production)"
).add_to(m)

folium.GeoJson(
    us_states,
    name="State details",
    style_function=lambda x: {"fillOpacity": 0, "color": "black", "weight": 0.3},
    popup=GeoJsonPopup(
        fields=["popup_html"],
        labels=False,              # don't print the field name
        parse_html=True
    ),
).add_to(m)

folium.LayerControl().add_to(m)

m

m.save("US_Total_Energy_Ratio_With_Mix.html")
```

```python
import folium
import requests
import numpy as np

# ---------------------------------
# 0. Build production/consumption shares
# ---------------------------------
energy_sources = {
    "Coal":        ("coal_production_2023",    "coal_consumption_2023"),
    "Natural Gas": ("ng_production_2023",      "ng_consumption_2023"),
    "Nuclear":     ("nuclear_production_2023", "nuclear_consumption_2023"),
    "Renewable":   ("renewable_production_2023","renewable_consumption_2023"),
    # "Petroleum": ("petroleum_production_2023","petroleum_consumption_2023")  # ← add if you have these
}

for label, (prod_col, cons_col) in energy_sources.items():
    if prod_col in merged_df.columns:
        merged_df[f"{label}_prod_share"] = (
            merged_df[prod_col] / merged_df["total_production_2023"].replace({0: np.nan}) * 100
        )
    if cons_col in merged_df.columns:
        merged_df[f"{label}_cons_share"] = (
            merged_df[cons_col] / merged_df["total_consumption_2023"].replace({0: np.nan}) * 100
        )

# ---------------------------------
# 1. Load US States GeoJSON
# ---------------------------------
geo_url = "https://raw.githubusercontent.com/python-visualization/folium/master/examples/data/us-states.json"
us_states = requests.get(geo_url).json()

# Use State as index for easy lookup
state_data = merged_df.set_index("State")

# ---------------------------------
# 2. Attach merged_df values + shares to GeoJSON features
# ---------------------------------
for feature in us_states["features"]:
    st = feature["id"]  # two-letter state code
    if st in state_data.index:
        row = state_data.loc[st]

        props = {
            "State": st,
            "Total Ratio (C/P)": row["total_ratio"],
            "Total Ratio (C/P) Capped": row["total_ratio_capped"],
            "Total Production": row["total_production_2023"],
            "Total Consumption": row["total_consumption_2023"],
        }

        # add production/consumption mix %
        for label in energy_sources.keys():
            prod_key = f"{label}_prod_share"
            cons_key = f"{label}_cons_share"
            if prod_key in row.index:
                props[f"{label} Prod %"] = row[prod_key]
            if cons_key in row.index:
                props[f"{label} Cons %"] = row[cons_key]

        feature["properties"].update(props)

# ---------------------------------
# 3. Create Choropleth Map (Leaflet)
# ---------------------------------
m = folium.Map(
    location=[37.8, -96],
    zoom_start=4,
    tiles="cartodbpositron"  # clean, accessible basemap
)

folium.Choropleth(
    geo_data=us_states,
    data=merged_df,
    columns=["State", "total_ratio_capped"],
    key_on="feature.id",
    fill_color="YlOrRd",      # colorblind-friendly palette
    fill_opacity=0.8,
    line_opacity=0.4,
    nan_fill_color="lightgray",
    legend_name="Total Energy Dependency Ratio (Consumption ÷ Production)"
).add_to(m)

# ---------------------------------
# 4. Add tooltips on hover (now including mix %)
# ---------------------------------
fields = [
    "State",
    "Total Ratio (C/P)",
    "Total Ratio (C/P) Capped",
    "Total Production",
    "Total Consumption",
]

aliases = [
    "State:",
    "True Ratio (Consumption ÷ Production):",
    "Capped Ratio (for map scale):",
    "Total Production (BBTU):",
    "Total Consumption (BBTU):",
]

# add dynamic energy mix fields
for label in energy_sources.keys():
    fields.extend([f"{label} Prod %", f"{label} Cons %"])
    aliases.extend([
        f"{label} Share of Production (%):",
        f"{label} Share of Consumption (%):",
    ])

tooltip = folium.features.GeoJsonTooltip(
    fields=fields,
    aliases=aliases,
    localize=True,
    sticky=True,
    labels=True,
)

folium.GeoJson(
    us_states,
    name="State Details",
    tooltip=tooltip,
    style_function=lambda x: {"fillOpacity": 0, "color": "black", "weight": 0.3}
).add_to(m)

folium.LayerControl().add_to(m)

# ---------------------------------
# 5. Display & Save
# ---------------------------------
m  # displays the map in Jupyter

m.save("US_Total_Energy_Ratio_Map_with_Mix.html")
print("Map saved as US_Total_Energy_Ratio_Map_with_Mix.html")
```

```python
import pandas as pd
import numpy as np
import folium
import requests

# ============================================
# 1. CLEAN NUMERIC COLUMNS (SAFE STEP)
# ============================================
num_cols = [c for c in merged_df.columns if c != "State"]

for col in num_cols:
    merged_df[col] = (
        merged_df[col]
        .astype(str)
        .str.strip()
        .str.replace(",", "", regex=False)
    )
    merged_df[col] = pd.to_numeric(merged_df[col], errors="coerce")

# ============================================
# 2. TOTAL RATIO (CONSUMPTION / PRODUCTION)
# ============================================
merged_df["total_ratio"] = (
    merged_df["total_consumption_2023"] /
    merged_df["total_production_2023"].replace({0: np.nan})
)

merged_df["total_ratio_capped"] = merged_df["total_ratio"].clip(upper=5)

# ============================================
# 3. CREATE "OTHER" CATEGORIES (RESIDUAL)
# ============================================
# Other Consumption = total - (coal + gas + nuclear + renewables)
merged_df["other_consumption_2023"] = (
    merged_df["total_consumption_2023"]
    - (
        merged_df["coal_consumption_2023"] +
        merged_df["ng_consumption_2023"] +
        merged_df["nuclear_consumption_2023"] +
        merged_df["renewable_consumption_2023"]
    )
).clip(lower=0)

# Other Production = total - (coal + gas + nuclear + renewables)
merged_df["other_production_2023"] = (
    merged_df["total_production_2023"]
    - (
        merged_df["coal_production_2023"] +
        merged_df["ng_production_2023"] +
        merged_df["nuclear_production_2023"] +
        merged_df["renewable_production_2023"]
    )
).clip(lower=0)

# ============================================
# 4. COMPUTE SHARES (% OF TOTAL) FOR EACH FUEL
# ============================================
# Consumption shares
merged_df["Coal Cons %"]        = merged_df["coal_consumption_2023"]      / merged_df["total_consumption_2023"] * 100
merged_df["Natural Gas Cons %"] = merged_df["ng_consumption_2023"]        / merged_df["total_consumption_2023"] * 100
merged_df["Nuclear Cons %"]     = merged_df["nuclear_consumption_2023"]   / merged_df["total_consumption_2023"] * 100
merged_df["Renewable Cons %"]   = merged_df["renewable_consumption_2023"] / merged_df["total_consumption_2023"] * 100
merged_df["Other Cons %"]       = merged_df["other_consumption_2023"]     / merged_df["total_consumption_2023"] * 100

# Production shares
merged_df["Coal Prod %"]        = merged_df["coal_production_2023"]       / merged_df["total_production_2023"] * 100
merged_df["Natural Gas Prod %"] = merged_df["ng_production_2023"]         / merged_df["total_production_2023"] * 100
merged_df["Nuclear Prod %"]     = merged_df["nuclear_production_2023"]    / merged_df["total_production_2023"] * 100
merged_df["Renewable Prod %"]   = merged_df["renewable_production_2023"]  / merged_df["total_production_2023"] * 100
merged_df["Other Prod %"]       = merged_df["other_production_2023"]      / merged_df["total_production_2023"] * 100

# ============================================
# 5. LOAD US STATES GEOJSON
# ============================================
geo_url = "https://raw.githubusercontent.com/python-visualization/folium/master/examples/data/us-states.json"
us_states = requests.get(geo_url).json()

state_data = merged_df.set_index("State")

# Attach all needed props to each state feature
for feature in us_states["features"]:
    st = feature["id"]  # state code
    if st in state_data.index:
        row = state_data.loc[st]
        feature["properties"].update({
            "State": st,
            "Total Ratio (C/P)": row["total_ratio"],
            "Total Ratio (C/P) Capped": row["total_ratio_capped"],
            "Total Production": row["total_production_2023"],
            "Total Consumption": row["total_consumption_2023"],
            "Coal Prod %":        row["Coal Prod %"],
            "Coal Cons %":        row["Coal Cons %"],
            "Natural Gas Prod %": row["Natural Gas Prod %"],
            "Natural Gas Cons %": row["Natural Gas Cons %"],
            "Nuclear Prod %":     row["Nuclear Prod %"],
            "Nuclear Cons %":     row["Nuclear Cons %"],
            "Renewable Prod %":   row["Renewable Prod %"],
            "Renewable Cons %":   row["Renewable Cons %"],
            "Other Prod %":       row["Other Prod %"],
            "Other Cons %":       row["Other Cons %"],
        })

# ============================================
# 6. BUILD CHOROPLETH MAP
# ============================================
m = folium.Map(
    location=[37.8, -96],
    zoom_start=4,
    tiles="cartodbpositron"
)

folium.Choropleth(
    geo_data=us_states,
    data=merged_df,
    columns=["State", "total_ratio_capped"],
    key_on="feature.id",
    fill_color="YlOrRd",
    fill_opacity=0.8,
    line_opacity=0.4,
    nan_fill_color="lightgray",
    legend_name="Total Energy Dependency Ratio (Consumption ÷ Production)"
).add_to(m)

# ============================================
# 7. TOOLTIP WITH FULL MIX INCLUDING "OTHER"
# ============================================
fields = [
    "State",
    "Total Ratio (C/P)",
    "Total Ratio (C/P) Capped",
    "Total Production",
    "Total Consumption",
    "Coal Prod %",
    "Coal Cons %",
    "Natural Gas Prod %",
    "Natural Gas Cons %",
    "Nuclear Prod %",
    "Nuclear Cons %",
    "Renewable Prod %",
    "Renewable Cons %",
    "Other Prod %",
    "Other Cons %",
]

aliases = [
    "State:",
    "True Ratio (Consumption ÷ Production):",
    "Capped Ratio (for map scale):",
    "Total Production (BBTU):",
    "Total Consumption (BBTU):",
    "Coal Share of Production (%):",
    "Coal Share of Consumption (%):",
    "Natural Gas Share of Production (%):",
    "Natural Gas Share of Consumption (%):",
    "Nuclear Share of Production (%):",
    "Nuclear Share of Consumption (%):",
    "Renewable Share of Production (%):",
    "Renewable Share of Consumption (%):",
    "Other Share of Production (%):",
    "Other Share of Consumption (%):",
]

tooltip = folium.features.GeoJsonTooltip(
    fields=fields,
    aliases=aliases,
    localize=True,
    sticky=True,
    labels=True,
)

folium.GeoJson(
    us_states,
    name="State Details",
    tooltip=tooltip,
    style_function=lambda x: {"fillOpacity": 0, "color": "black", "weight": 0.3},
).add_to(m)

folium.LayerControl().add_to(m)

# ============================================
# 8. DISPLAY & SAVE
# ============================================
m  # display in Jupyter

m.save("US_Total_Energy_Ratio_Map_with_Other.html")
print("Map saved as US_Total_Energy_Ratio_Map_with_Other.html")
```
